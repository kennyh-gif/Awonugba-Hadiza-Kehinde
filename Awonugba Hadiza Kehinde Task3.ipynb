```python
# changing the date to datetime
pizzas['date']=pd.to_datetime(pizzas['date'])
```

**1. What is the total revenue/sales?**

```python
# creating a new column to calculate total sales/revenue
pizzas['revenue']=pizzas['price']*pizzas['quantity']
```

```python
# this is the total sales made(total revenue)
total_sales= pizzas['revenue'].sum()
print(f'The total sales/revenue is {total_sales:,.2f}')
```

**2. Find the total quantity sold.**

```python
total_q= pizzas['quantity'].sum()
print(f"The total quantity of pizza sold is {total_q} pizzas")
```

**3. Find the total orders**

```python
# counting only the unique values in the order_id
p_orders= pizzas['order_id'].nunique()
print(f"The total number of orders is {p_orders}")
```

**4. How many pizza types do they sell?**

```python
# listing the different types of pizzas
p=pizzas['name'].value_counts()
```

```python
# counting how many pizzas from the list above
pizza_types=p.count()
print(f'They sell {pizza_types} different types of pizzas')
```

**5. Find the average price of pizza**

```python
avg_price=pizzas['price'].mean()
print(f"The average price of pizza is {avg_price:.2f}")
```

```python
# creating a new column for month from the date column
pizzas['months']=pizzas['date'].dt.month_name()
```

```python
# creating a new column for days of the weeek from the date column
pizzas['days']=pizzas['date'].dt.day_name()
```

```python
# creating a new column for hours of the day from the time column
pizzas['time(hours)']=pizzas['time'].dt.hour
```

```python
pizzas
```

```python
pizzas['time']=pizzas['time'].dt.time #removing date from the time column
```

```python
pizzas
```

**6. What are the peak hours of sales?**

```python
# getting the number of sales made each hour
sales_per_hour=pizzas.groupby('time(hours)')['revenue'].sum().reset_index()
sales_per_hour
```

```python
plt.plot(sales_per_hour['time(hours)'],sales_per_hour['revenue'], marker='.')
plt.title("Total Sales Per Hour",fontdict={'fontweight':700,'fontname':'Times New Roman','fontsize':24})
plt.xlabel("Hours")
plt.ylabel("Total Sales")
plt.grid(alpha=0.5)

plt.show()
```

### **From the plot above the peak hours of sales are the 12th and 13th hours(noon)** ###

**7. Find the total sales made on each day of the week. Which day of the week is when sales are made the most?**

```python
# grouping by days of the week ang getting the total revenue for each day
sales_per_day= pizzas.groupby('days')['revenue'].sum().reset_index()
```

```python
# arranging the days in the correct order
day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday","Sunday"]

sales_per_day['days'] = pd.Categorical(sales_per_day['days'], categories=day_order, ordered=True)
```

```python
sales_per_day=sales_per_day.sort_values('days')
```

```python
print(f"The sale made on each day of the week: ")
sales_per_day
```

```python
plt.bar(sales_per_day['days'],sales_per_day['revenue'], color='#abc')
plt.title("Total Sales Per Day",fontdict={'fontweight':700,'fontname':'Times New Roman','fontsize':24})
plt.xlabel("Days Of The Week")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.show()
```

### **From the plot above, sales are made the most on fridays** ###

**9. Find the sales made in each month. Any trend noticeable?**

```python
# grouping by months and getting the total revenue for each month
sales_per_month= pizzas.groupby('months')['revenue'].sum().reset_index()
```

```python
# arranging the months in the correct order
month_order = ["January", "February", "March", "April", "May", "June", 
               "July", "August", "September", "October", "November", "December"]
# changing the months column to the correct order
sales_per_month['months'] = pd.Categorical(sales_per_month['months'], categories=month_order, ordered=True)
```

```python
sales_per_month=sales_per_month.sort_values('months')
```

```python
sales_per_month
```

```python
plt.plot(sales_per_month['months'],sales_per_month['revenue'], marker='.')
plt.title("Total Sales Per Month",fontdict={'fontweight':700,'fontname':'Times New Roman','fontsize':24})
plt.xlabel("Months")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

### **Trend: Sales are fairly steady, but there are spikes around March to July.**
### **Overall, sales don’t show a big upward or downward trend — more of a seasonal fluctuation pattern.** ###
